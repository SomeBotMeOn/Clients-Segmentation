<strong><center><h1> Agglomerative Clustering</strong>

```python
import pandas as pd
import scipy.cluster.hierarchy as sch
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering
from sklearn.preprocessing import StandardScaler
```

```python
data_path = '../../../../data/processed_data/main_datasets/customer_data_processed.csv'
data = pd.read_csv(data_path)

X = data.values

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

plt.figure(figsize=(10, 7))
sch.dendrogram(sch.linkage(X_scaled, method='ward'))

plt.title('Дендограмма')
plt.xlabel('Индекс образца')
plt.ylabel('Расстояние')

plt.show()
```

Как мы видим, наиболее подходящее количество кластеров равно 3

```python
n_clusters = 3 
agg_clustering = AgglomerativeClustering(n_clusters=n_clusters)

data['Cluster'] = agg_clustering.fit_predict(X_scaled)
```

```python
output_path = ('../../../../data/processed_data/train_models_data/data_without_PCA/data_without_generated_features/AgglomerativeClustering_default.csv')
data.to_csv(output_path, index=False)
```
