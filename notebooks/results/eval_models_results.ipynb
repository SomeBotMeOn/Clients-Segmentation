<h1><center><strong> Eval Models Results</center></h1>

Мы использовали методы кластеризации для создания различных датасетов, а затем проверяли их с помощью Random Forest, чтобы определить наилучший подбор кластеров. В результате мы получили два претендента: **DBSCAN_generated_features_PCA** с точностью 0.99344 и **AgglomerativeClustering_generated_features** с точностью 0.99075. Теперь нам необходимо более подробно изучить, как именно распределены данные между кластерами в обоих случаях, чтобы понять, какой из алгоритмов кластеризации более эффективно разделяет данные.

<center><strong>DBSCAN_generated_features_PCA</center>

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.metrics.pairwise import euclidean_distances

customer_data = pd.read_csv('../../data/raw_data/customer_data.csv')  
dbscan_data = pd.read_csv('../../data/processed_data/train_models_data/data_with_PCA/data_with_generated_features/DBSCAN_generated_features_PCA.csv') 

features = dbscan_data.drop(columns=['Cluster'])

scaler = StandardScaler()
features_scaled = scaler.fit_transform(features)

clusters = dbscan_data['Cluster'].unique()

closest_points = []

for cluster in clusters:
    cluster_data = features_scaled[dbscan_data['Cluster'] == cluster]
    
    cluster_center = np.mean(cluster_data, axis=0)
    
    distances = euclidean_distances(cluster_data, cluster_center.reshape(1, -1))
    closest_point_index = np.argmin(distances)
    
    customer_index = dbscan_data[dbscan_data['Cluster'] == cluster].index[closest_point_index]
    closest_point = customer_data.iloc[customer_index]
    
    closest_points.append(closest_point)

result_df = pd.DataFrame(closest_points)
result_df
```

<center><strong>AgglomerativeClustering_generated_features</center>

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.metrics.pairwise import euclidean_distances

customer_data = pd.read_csv('../../data/raw_data/customer_data.csv')  
agg_clustering_data = pd.read_csv('../../data/processed_data/train_models_data/data_without_PCA/data_with_generated_features/AgglomerativeClustering_generated_features.csv') 

features = agg_clustering_data.drop(columns=['Cluster'])

scaler = StandardScaler()
features_scaled = scaler.fit_transform(features)

clusters = agg_clustering_data['Cluster'].unique()

closest_points = []

for cluster in clusters:
    cluster_data = features_scaled[agg_clustering_data['Cluster'] == cluster]
    
    cluster_center = np.mean(cluster_data, axis=0)
    
    distances = euclidean_distances(cluster_data, cluster_center.reshape(1, -1))
    closest_point_index = np.argmin(distances)
    
    customer_index = agg_clustering_data[agg_clustering_data['Cluster'] == cluster].index[closest_point_index]
    closest_point = customer_data.iloc[customer_index]
    
    closest_points.append(closest_point)

result_df = pd.DataFrame(closest_points)
result_df
```

#### Выводы: можем заметить, что AgglomerativeClustering лучше разделяет данные, так как кластеры более логичные и видима существенная разница между ними с точки зрения экономической логики. В то время как DBSCAN разделил данные на кластеры, которые сложно интерпретировать и практически не имеют смысла.
