### Посмотрим, влияет ли наличие дополнительных признаков на качество модели

```python
import pandas as pd
from itertools import combinations_with_replacement
```

```python
data = pd.read_csv('../../data/processed_data/main_datasets/customer_data_processed.csv')
data.head(5)
```

```python
def calc_stats(col):
    return {
        'mean': col.mean(),
        'std': col.std(),
        'min': col.min(),
        'max': col.max(),
        'skew': col.skew(),
        'median': col.median()
    }

for col1, col2 in combinations_with_replacement(data.columns, 2):
    # Операции над колонками
    if col1 != col2:  # Для суммы - без повторений
        data[f'{col1}_plus_{col2}'] = data[col1] + data[col2]
        data[f'{col1}_proizvedenie_{col2}'] = data[col1] * data[col2]
    
    # Для разности - с повторениями
    data[f'{col1}_minus_{col2}'] = data[col1] - data[col2]

for col in data.columns:
    stats = calc_stats(data[col])
    for stat, value in stats.items():
        data[f'{col}_{stat}'] = value

data
```
